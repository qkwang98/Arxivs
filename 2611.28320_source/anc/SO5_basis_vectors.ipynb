Below are some sanity check methods to see that a purported highest weight vector in the list above behaves as it should. Taking the Laplacian checks that the vector's coefficients are harmonic polynomials. Computing its behaviour under action by the basis of the Cartan subalgebra $T_1$ and $T_2$ checks that the vector is of the correct weight. It is a highest weight vector if and only if it is annihilated by all postive roots. Finally, it is verified that the chosen vector is orthogonal to the other highest weight vectors in the list.

```python
family_name = 'A'
degree = 1
highest_weight = (2,0)

family = globals()[family_name]
Psi = sp.simplify(family[degree][highest_weight][highest_weight][0])

display(sp.simplify(form_Laplacian(Psi)))

for m in [T1,T2]: display(proportionality(Psi,act_on_form(m,Psi)))

for ind in [(0,1),(1,1),(1,0),(1,-1)]: display(sp.factor(act_on_form(adRep[ind],Psi)))

lower_degree_forms = sum([
        list(map(lambda Psi: normsquared**k*Psi,
            ((A[degree-2*k][highest_weight][highest_weight] if highest_weight in A[degree-2*k] else []) if degree-2*k in A else [])+((B[degree-2*k][highest_weight][highest_weight] if highest_weight in B[degree-2*k] else []) if degree-2*k in B else [])+((C[degree-2*k][highest_weight][highest_weight] if highest_weight in C[degree-2*k] else []) if degree-2*k in C else [])
        ))
    for k in range(0,3)],[])
for m in lower_degree_forms: display(form_inner_product(Psi,m))
```
