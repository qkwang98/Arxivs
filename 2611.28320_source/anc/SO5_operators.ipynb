The following cells contain the definition of various $SO(5)$-equivariant linear operators constituting the linearized Yang--Mills operator $\mathbf L$. In terms of the notation from the accompanying paper, $\mathbf T_1$ is `diff_W`, $\mathbf T_2$ is `double_commutator` and $\mathbf T_3, \mathbf T_4, \mathbf T_5$ are `curvature_commutator_A`, `curvature_commutator_B` and `curvature_commutator_C`, respectively.

```python
# An SO(5)-invariant differential operator given by F -> [W_j,[W_j,F_i]]dx^i, using Einstein's summation convention.
def double_commutator(Psi):
    Psi_int = [sp.Matrix(Psi[i]) for i in range(sp.shape(Psi)[0])]
    a = [sp.Matrix(A[1][0,0][0,0][0][i]) for i in range(sp.shape(Psi)[0])]
    return sp.Array([sum([a[j]*a[j]*Psi_int[i]+Psi_int[i]*a[j]*a[j]-2*a[j]*Psi_int[i]*a[j]
                           for j in range(sp.shape(Psi)[0])],sp.zeros(sp.shape(Psi)[0])).tolist()
                     for i in range(sp.shape(Psi)[0])])
```

```python
# An SO(5)-invariant differential operator given by F -> [W_j,d_j F_i] dx^i, using Einstein's summation convention.
def diff_W(Psi):
    Psi_int = [sp.Matrix(Psi[i]) for i in range(sp.shape(Psi)[0])]
    a = [sp.Matrix(A[1][0,0][0,0][0][i]) for i in range(sp.shape(Psi)[0])]
    return sp.Array([sum([a[j]*sp.diff(Psi_int[i],x[j])-sp.diff(Psi_int[i],x[j])*a[j]
                           for j in range(sp.shape(Psi)[0])],sp.zeros(sp.shape(Psi)[0])).tolist()
                     for i in range(sp.shape(Psi)[0])])
```

```python
# An SO(5)-invariant differential operator given by F -> [F_j,d_i W_j - d_j W_i - [W_i,W_j]] dx^i, using Einstein's summation convention.
def curvature_commutator(Psi):
    Psi_int = [sp.Matrix(Psi[i]) for i in range(sp.shape(Psi)[0])]
    a = [sp.Matrix(A[1][0,0][0,0][0][i]) for i in range(sp.shape(Psi)[0])]
    curvature = [[sp.diff(a[j],x[i]) - sp.diff(a[i],x[j])  - (a[i]*a[j]-a[j]*a[i]) for i in range(sp.shape(Psi)[0])] for j in range(sp.shape(Psi)[0])]
    return sp.Array([sum([Psi_int[i]*curvature[j][i]-curvature[j][i]*Psi_int[i] for i in range(sp.shape(Psi)[0])],sp.zeros(sp.shape(Psi)[0])).tolist()
            for j in range(sp.shape(Psi)[0])])
```

```python
def curvature_commutator_A(Psi):
    Psi_int = [sp.Matrix(Psi[i]) for i in range(sp.shape(Psi)[0])]
    a = [sp.Matrix(A[1][0,0][0,0][0][i]) for i in range(sp.shape(Psi)[0])]
    curvature = [[sp.diff(a[j],x[i]) - sp.diff(a[i],x[j]) for i in range(sp.shape(Psi)[0])] for j in range(sp.shape(Psi)[0])]
    return sp.Array([sum([Psi_int[i]*curvature[j][i]-curvature[j][i]*Psi_int[i] for i in range(sp.shape(Psi)[0])],sp.zeros(sp.shape(Psi)[0])).tolist()
            for j in range(sp.shape(Psi)[0])])
```

```python
def curvature_commutator_B(Psi):
    Psi_int = [sp.Matrix(Psi[i]) for i in range(sp.shape(Psi)[0])]
    a = [sp.Matrix(A[1][0,0][0,0][0][i]) for i in range(sp.shape(Psi)[0])]
    curvature = [[x[i]*a[j] - x[j]*a[i] for i in range(sp.shape(Psi)[0])] for j in range(sp.shape(Psi)[0])]
    return sp.Array([sum([Psi_int[i]*curvature[j][i]-curvature[j][i]*Psi_int[i] for i in range(sp.shape(Psi)[0])],sp.zeros(sp.shape(Psi)[0])).tolist()
            for j in range(sp.shape(Psi)[0])])
```

```python
def curvature_commutator_C(Psi):
    Psi_int = [sp.Matrix(Psi[i]) for i in range(sp.shape(Psi)[0])]
    a = [sp.Matrix(A[1][0,0][0,0][0][i]) for i in range(sp.shape(Psi)[0])]
    curvature = [[(a[i]*a[j]-a[j]*a[i]) for i in range(sp.shape(Psi)[0])] for j in range(sp.shape(Psi)[0])]
    return sp.Array([sum([Psi_int[i]*curvature[j][i]-curvature[j][i]*Psi_int[i] for i in range(sp.shape(Psi)[0])],sp.zeros(sp.shape(Psi)[0])).tolist()
            for j in range(sp.shape(Psi)[0])])
```

```python
def form_rdr(Psi):
    Psi_int = [sp.Matrix(Psi[i]) for i in range(sp.shape(Psi)[0])]
    return sp.Array([sum([x[i]*sp.diff(Psi_int[j],x[i]) for i in range(sp.shape(Psi)[0])],sp.zeros(sp.shape(Psi)[0])).tolist()
            for j in range(sp.shape(Psi)[0])])
```

```python
def form_Laplacian(Psi):
    Psi_int = [sp.Matrix(Psi[i]) for i in range(sp.shape(Psi)[0])]
    return sp.Array([sum([sp.diff(Psi_int[j],(x[i],2)) for i in range(sp.shape(Psi)[0])],sp.zeros(sp.shape(Psi)[0])).tolist()
            for j in range(sp.shape(Psi)[0])])
```

```python
def form_angular_Laplacian(Psi):
    dimension = 5
    normsquared = (x[0]**2+x[1]**2+x[2]**2+x[3]**2+x[4]**2)
    Psi_int = [sp.Matrix(Psi[i]) for i in range(sp.shape(Psi)[0])]
    laplace_Psi = sp.Array([sum([sp.diff(Psi_int[j],(x[i],2)) for i in range(sp.shape(Psi)[0])],sp.zeros(sp.shape(Psi)[0])).tolist()
            for j in range(sp.shape(Psi)[0])])
    return laplace_Psi - form_rdr(form_rdr(Psi)) - (dimension-2) * form_rdr(Psi)
```

```python
def curvature_form(Psi):
    Psi_int = [sp.Matrix(Psi[i]) for i in range(sp.shape(Psi)[0])]
    curvature = [[sp.diff(Psi_int[j],x[i]) - sp.diff(Psi_int[i],x[j])  + (Psi_int[i]*Psi_int[j]-Psi_int[j]*Psi_int[i]) for i in range(sp.shape(Psi)[0])] for j in range(sp.shape(Psi)[0])]
    return curvature
```

The following two cells contain two instances of the entire operator $\mathbf L$. The first instance is line-by-line identical to the definition from the accompanying paper, while the second one references its decomposition into the constituent operators $\mathbf T_1,\ldots,\mathbf T_5$. One can check that the two yield identical results to rule out bugs in the definition of $\mathbf T_1,\ldots,\mathbf T_5$ in this file.

```python
def full_linear_operator(Psi):
    Psi_int = [sp.Matrix(Psi[i]) for i in range(sp.shape(Psi)[0])]
    a = sp.sqrt(6)/4
    b = sp.simplify(sp.Integer(9) - sp.Integer(7)*sp.sqrt(6)/2)
    W = sp.Array([(A[1][0,0][0,0][0][j]/(a*normsquared+b)).tolist() for j in range(5)])
    Omega = curvature_form(W)
    comm_Omega = sp.Array([sum([Psi_int[i]*Omega[j][i]-Omega[j][i]*Psi_int[i] for i in range(5)],sp.zeros(5)).tolist() for j in range(5)])
    return form_Laplacian(Psi)-1/2*form_rdr(Psi)-1/2*Psi+2/(a*normsquared+b)*diff_W(Psi)+1/(a*normsquared+b)**2*double_commutator(Psi)+2*comm_Omega
```

```python
def reduced_linear_operator(Psi):
    Psi_int = [sp.Matrix(Psi[i]) for i in range(sp.shape(Psi)[0])]
    a = sp.sqrt(6)/4
    b = sp.simplify(sp.Integer(9) - sp.Integer(7)*sp.sqrt(6)/2)
    W = sp.Array([(A[1][0,0][0,0][0][j]/(a*normsquared+b)).tolist() for j in range(5)])
    Omega = curvature_form(W)
    comm_Omega = sp.Array([sum([Psi_int[i]*Omega[j][i]-Omega[j][i]*Psi_int[i] for i in range(5)],sp.zeros(5)).tolist() for j in range(5)])
    return (form_Laplacian(Psi)-1/2*form_rdr(Psi)-1/2*Psi
            +2/(a*normsquared+b)*diff_W(Psi)+1/(a*normsquared+b)**2*double_commutator(Psi)
            +2/(a*normsquared+b)**2*(-2*a*curvature_commutator_C(Psi)+curvature_commutator_B(Psi)+(a*normsquared+b)*curvature_commutator_A(Psi)))
```

The following cell contains the bulk of the computational work: Each matrix $T_j$ is computed by applying the operator $\mathbf T_j$ to each basis element and then taking the normalized inner product
$$ \left\langle v_k, T_j v_{k'} \right\rangle \| v_k \|^{-1} \| v_{k'} \|^{-1}.$$

```python
t1 = get_coeffs_of_invariant_operator(diff_W,basis) #has weight f, scales with 1
t2 = get_coeffs_of_invariant_operator(double_commutator,basis) #has weight f^2, scales with r^2
t3 = get_coeffs_of_invariant_operator(curvature_commutator_A,basis) #has weight f, scales with 1
t4 = get_coeffs_of_invariant_operator(curvature_commutator_B,basis) #has weight f', scales with r^2
t5 = get_coeffs_of_invariant_operator(curvature_commutator_C,basis) #has weight f^2, scales with r^2
#t6 = get_coeffs_of_invariant_operator(form_angular_Laplacian,basis)
```
